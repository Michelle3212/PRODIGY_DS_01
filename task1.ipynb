```python
import pandas as pd
df = pd.read_csv("world_country_stats.csv")
df.head()
```

```python
df.tail()
```

```python
df.columns
```

```python
df = df.dropna()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plot
```

```python
cat_coun = df['region'].value_counts()
```

```python
cat_coun
```

```python
plot.figure()
cat_coun.plot(kind='bar')

plot.title("Distribution by Region")
plot.xlabel("Region")
plot.ylabel("Count")
plot.show()
```

```python
plot.figure()

plot.hist(df['median_age'], bins=20)

plot.title("Distribution of Median Age")
plot.xlabel("Median Age")
plot.ylabel("Frequency")

plot.show()
```

```python
plot.figure()

plot.hist(df['fertility_rate'], bins=20)

plot.title("Distribution of Fertility Rate")
plot.xlabel("Fertility Rate")
plot.ylabel("Frequency")

plot.show()
```

```python
plot.figure()

plot.hist(df['land_area'], bins=20)

plot.title("Distribution of Land Area")
plot.xlabel("Land Area")
plot.ylabel("Frequency")

plot.show()
```

Bar charts were used to visualize categorical distribution of countries by region, while histograms were used to analyze continuous variables such as median age, fertility rate, and land area. Log transformation was applied to land area to handle skewness.
